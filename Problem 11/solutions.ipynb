```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('50_Startups.csv')
```

```python
data.head()
```

```python
data.info()
```

```python
features = data.iloc[:,[0,1,2,3]].values
label = data.iloc[:,4].values
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
oheState = OneHotEncoder(sparse_output=True)

dummyState = oheState.fit_transform(features[:,3].reshape(-1,1)).toarray()
```

```python
dummyState
```

```python
final_features = np.concatenate((dummyState,features[:,[0,1,2]]),axis=1)
```

```python
final_features
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
```

```python
for i in range(1,301):
    for depth in range(3,11):
        x_train,x_test,y_train,y_test = train_test_split(final_features,label,test_size=0.2,random_state=i)
        
        model = RandomForestRegressor(
            n_estimators=5,
            max_depth=depth,
            random_state=2
        )
        
        model.fit(x_train,y_train)
        
        trainScore = model.score(x_train,y_train)
        testScore = model.score(x_test,y_test)
        
        if testScore>trainScore and testScore >=0.9:
            print(f"Test {testScore} | Train {trainScore} and rs {i} | depth {depth}")
        
```

```python
# Test 0.9818264162381104 | Train 0.9773719985661841 and rs 158 | depth 5
x_train,x_test,y_train,y_test = train_test_split(final_features,label,test_size=0.2,random_state=158)
model = RandomForestRegressor(
    n_estimators=5,
    max_depth=5,
    random_state=2
)

model.fit(x_train,y_train)
```
