```python
import pandas as pd 
import numpy as np
```

```python
data = pd.read_csv('50_Startups.csv')
```

```python
data.info()
```

```python
features = data.iloc[:,[0,1,2,3]].values
label = data.iloc[:,[4]].values
```

```python
from sklearn.preprocessing import OneHotEncoder

oheState = OneHotEncoder(sparse_output=False)

stateDummmy = oheState.fit_transform(features[:,[3]])
```

```python
finalFeature = np.concatenate((stateDummmy,features[:,[0,1,2]]),axis=1)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
CL = 0.9

for rs in range(0,101):
    x_train,x_test,y_train,y_test = train_test_split(finalFeature,label,test_size=0.2,random_state=rs)
    model = LinearRegression()
    
    model.fit(x_train,y_train)
    
    trainScore = model.score(x_train,y_train)
    testScore = model.score(x_test,y_test)
    
    if testScore > trainScore and testScore>CL :
        print(f"Test Score : {testScore} and Train Score : {trainScore} , random state : {rs}")
```

```python
x_train , x_test , y_train , y_test = train_test_split(finalFeature,label,test_size=0.2,random_state=10)

model = LinearRegression()

model.fit(x_train,y_train)

print(f"Test score : {model.score(x_test,y_test)} and Train score : {model.score(x_train,y_train)}")
```

```python
oheState.categories_[0]
```

```python
rdSpend = float(input("Enter rd spend "))
adminSpend = float(input("Enter admin spend "))
marketSpend = float(input("Enter market spend "))
state = input("Enter state ")
state = str.capitalize(state)
print(state)
if state in oheState.categories_[0]:
    dummyV = oheState.transform(np.array([[state]]))
    print(f"Dummy state {dummyV}")
    finalFeatureInput = np.concatenate((dummyV,np.array([[rdSpend,adminSpend,marketSpend]])),axis=1)
    profit = model.predict(finalFeatureInput)
    print(f"Profit : {profit}")
else:
    print(f"Given State {state} is not recognized by AI model")
```
