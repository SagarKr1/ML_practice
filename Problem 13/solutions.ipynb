```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('50_Startups.csv')
```

```python
data.info()
```

```python
data.head()
```

```python
features = data.iloc[:,[0,1,2,3]].values
label = data.iloc[:,4].values
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
oheState = OneHotEncoder(sparse_output=True)
dummyState = oheState.fit_transform(features[:,3].reshape(-1,1)).toarray()
```

```python
finalFeature = np.concatenate((dummyState,features[:,[0,1,2]]),axis=1)
```

```python
from sklearn.model_selection import train_test_split
from xgboost import XGBRegressor
```

```python
x_train,x_test,y_train,y_test = train_test_split(finalFeature,label,test_size=0.2,random_state=62)
model = XGBRegressor(
    n_estimators=5,
    max_depth=7,
    random_state=10
)

model.fit(x_train,y_train)

trainScore = model.score(x_train,y_train)
testScore = model.score(x_test,y_test)

if testScore > 0.9:
    print(f"Test Score {testScore} and Train Score {trainScore}. Approved")
else :
    print(f"Test Score {testScore} and Train Score {trainScore} Improve Model")
```
