# Import Libraries

```python
import pandas as pd
import numpy as np
```

# Load Dataset

```python
data = pd.read_csv('dataset.csv')
```

# Basic info

```python
data.info()
```

```python
data.head()
```

# Seprate Feature and label

```python
features = data.iloc[:,[0,1]].values
label = data.iloc[:,2].values
```

# Feature Selection

```python
from sklearn.feature_selection import RFE
from sklearn.linear_model import LogisticRegression

algo = LogisticRegression()
```

```python
rfe = RFE(estimator=algo)

rfe.fit(features,label)
```

```python
rfe.ranking_
```

# Model Building and Metrics Evaluation

```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
```

```python
for rs in range(1,101):
    x_train,x_test,y_train,y_test = train_test_split(features,label,test_size=0.2,random_state=rs)
    
    for k in range(1,11):
        model = KNeighborsClassifier(n_neighbors=k)
    
        model.fit(x_train,y_train)
    
        trainScore = model.score(x_train,y_train)
        testScore = model.score(x_test,y_test)
        
        if testScore >= 0.8:
            print(f"Test Score {testScore} and Train Score {trainScore} , rs {rs} , k {k}")
```

```python
# Test Score 0.8333333333333334 and Train Score 0.8 , rs 2 , k 10
x_train,x_test,y_train,y_test = train_test_split(features,label,test_size=0.2,random_state=1)

model = KNeighborsClassifier(n_neighbors=1)

model.fit(x_train,y_train)

trainScore = model.score(x_train,y_train)
testScore = model.score(x_test,y_test)

if testScore >= 0.8:
    print("Model Approved")
else :
    print(f"Model need to imporve {testScore}")
```

```python
from sklearn.metrics import confusion_matrix

confusion_matrix(label,model.predict(features))
```

```python
model.predict(features) # Predicted Label
```

```python
label.ravel() # Actual label
```

```python
from sklearn.metrics import classification_report

print(classification_report(label,model.predict(features)))
```

```python
CL = 0.8

from sklearn.metrics import f1_score

f1ScoreFromClassificationReport = f1_score(label,model.predict(features),average='macro')
print(f1ScoreFromClassificationReport)
if f1ScoreFromClassificationReport >= CL:
    print("Model Approved")
else:
    print("Improve Model")
```
