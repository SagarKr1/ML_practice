```python
import pandas as pd 
import numpy as np
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.compose import ColumnTransformer
```

```python
data = pd.read_csv('student_performance_regression_dataset.csv')
```

```python
features = data.iloc[:,[0,1,2,3,4,5]].values
label = data.iloc[:,[6]].values
```

```python
preprocessingPipe  = ColumnTransformer([
    (
        "Numerical Imputer",
        Pipeline([
        ("Impute",SimpleImputer(strategy="mean"))
    ]),
    [0,1,2,3,4]
        ),
    (
        "Categorical Imputer",
        Pipeline([
        ("Impute",SimpleImputer(strategy="most_frequent"))
    ]),
        [5]
        )
])
```

```python
transformFeatures = preprocessingPipe.fit_transform(features)
transformFeatures
```

```python
pd.DataFrame(transformFeatures).isnull().sum()
```

```python
ready_finalFeature = ColumnTransformer([
    (
        "One Hot Encoding",
        Pipeline([
            ("OHE",OneHotEncoder()),
        ]),
        [5]
    ),
    (
        "Standard Scaler",
        Pipeline([
            ("Scaler",StandardScaler()),
        ]),
        [0,1,2,3,4]
    )
])
```

```python
finalFeature = ready_finalFeature.fit_transform(transformFeatures)
```

```python
ohe_columns = ready_finalFeature.named_transformers_[
    "One Hot Encoding"
].named_steps["OHE"].get_feature_names_out()
ohe_columns
```

```python
finalFeature.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(finalFeature,label,test_size=0.2,random_state=42)
```

```python
model = KNeighborsRegressor(n_neighbors=3)
```
